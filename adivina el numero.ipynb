```python
print(
    "Hola! Bienvenidos al juego de 'Adivina el número'. Como bien indica el título de este juego,\n"
    "el objetivo de este será adivinar un número generado aleatoriamente por la máquina entre el 1 y el 1000.\n\n"
    "Habrá tres dificultades:\n"
    "  1) Nivel fácil (20 intentos)\n"
    "  2) Nivel intermedio (12 intentos)\n"
    "  3) Nivel difícil (5 intentos)\n\n"
    "También disponemos del modo de juego solitario o en pareja.\n"
    "Además, se creará un archivo Excel donde se guardarán las estadísticas "
    "dependiendo de los diferentes modos de juego."
)
```

Selección de modo de juego.

```python
dificultad=0
modo='Nulo'
while dificultad not in [1,2,3] or modo not in ['Solitario', 'Pareja']:    
    dificultad=int(input('Dime con que dificultad quieres jugar.'))
    modo=input('Escribe "Solitario" para jugar solo, o "Pareja" para jugar en pareja')

if dificultad == 1 and modo == 'Solitario':
    print("Has elejido la dificultad fácil. Tienes 20 intentos.")
    numero_intentos=20
if dificultad == 1 and modo == 'Pareja':
    print("Habéis elejido la dificultad fácil. Teneis 20 intentos.")
    numero_intentos=20
elif dificultad == 2 and modo == 'Solitario':
    print("Has elejido la dificultad intermedia. Tienes 12 intentos.")
    numero_intentos=12
elif dificultad == 2 and modo == 'Pareja':
    print("Habéis elejido la dificultad intermedia. Teneis 12 intentos.")
    numero_intentos=12
elif dificultad == 3 and modo == 'Solitario':
    print("Habéis elejido la dificultad dificil. Tiene 5 intentos.")
    numero_intentos=5
elif dificultad == 3 and modo == 'Pareja':
    print("Habéis elejido la dificultad difícil. Teneis 5 intentos.")
    numero_intentos=5
            
```

Segun si es el modo 'Solitario' o en 'Pareja' pediremos un o dos nombres.

```python
if modo == 'Solitario':
    nombre = input('¿Cómo te llamas?')
if modo == 'Pareja':
    nombre_1 = input('¿Cómo se llama el primer integrante?')    
    nombre_2 = input('¿Cómo se llama el segundo integrante?')  
    nombres=[nombre_1, nombre_2] 
```

Empezamos a jugar.

```python
encierto = False  #Esta variable nos indica que no se ha encontrado el num secreto. Si se encuentra, encierto pasara a ser True y se romperá el bucle siguiente.
numero_elegido = random.randint(1,1001) #Se genera un número aleatorio entre el 1 y el 1000.
i = 0  #Es el contador.
if modo == 'Solitario':
    while i in range(numero_intentos) and encierto == False:
        intento=int(input(f'Tu intento número {i+1} es:'))
        if intento == numero_elegido:
            encierto=True
            print(f'Has acertado en tu intento {i+1}! El número secreto era {numero_elegido}!')
        if intento<numero_elegido:
            print(f'El número que me has dado es menor que el númerop secreto... Te quedan {numero_intentos-(i+1)} intentos!')
            i += 1
        if intento>numero_elegido:
            print(f'El número que me has dado es mayor que el númerop secreto... Te quedan {numero_intentos-(i+1)} intentos!')
            i += 1
    if encierto == False:
        print("Lo siento, no has adivinado el número secreto... Inténtalo otra vez!")

# Hasta aqui el juego en solitario. Ahora viene el juego en parejas.
j=0
if modo == 'Pareja':
    while j in range(numero_intentos) and encierto == False:
        intento_1=int(getpass(f'El intento número {j+1} de {nombre_1} es:'))
        if intento_1 == numero_elegido:
            print(f'Enhorabuena {nombre_1}! Has acertado en tu intento {j+1}! El número secreto era {numero_elegido}!')
            ganador=nombre_1
            perdedor=nombre_2
            break
        if intento_1<numero_elegido:
            print(f'El número que me has dado es menor que el númerop secreto... Te quedan {numero_intentos-(j+1)} intentos!')
        if intento_1>numero_elegido:
            print(f'El número que me has dado es mayor que el númerop secreto... Te quedan {numero_intentos-(j+1)} intentos!')
        intento_2=int(getpass(f'El intento número {j+1} de {nombre_2} es:'))
        if intento_2 == numero_elegido:
            print(f'Enhorabuena {nombre_2}! Has acertado en tu intento {i+1}! El número secreto era {numero_elegido}!')
            ganador = nombre_2
            perdedor = nombre_1
            break
        if intento_2<numero_elegido:
            print(f'El número que me has dado es menor que el númerop secreto... Te quedan {numero_intentos-(i+1)} intentos!')
            j += 1
        if intento_2>numero_elegido:
            print(f'El número que me has dado es mayor que el númerop secreto... Te quedan {numero_intentos-(i+1)} intentos!')
            j += 1
        

```
